**Q10. Handling Missing values**

```python
#Considering the following DataFrame
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Age': [21, np.nan, 23, 22],
    'Score': [80, 75, np.nan, 90]})

#Identify missing values
print("Missing values in the df")
print(df.isna().sum())

#Replacing the missing age with mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())

#Replacing the missing score with mean score
df['Score'] = df['Score'].fillna(df['Score'].mean())

print("\nDataFrame without missing values")
df
```

**Q13. Feature Creation**

```python
df_sp['Average_Technical_Score'] = df_sp[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)

def get_performance(score):
    if score >= 90:
        return 'Excellent'
    elif score >= 75:
        return 'Good'
    elif score >= 60:
        return 'Average'
    else:
        return 'Needs Improvement'

df_sp['Performance'] = df_sp['Average_Technical_Score'].apply(get_performance)

df_sp[['Name', 'Average_Technical_Score', 'Performance']].head()
```

**Q14. GroupBy Analysis**

```python
dept_summary = df_sp.groupby('Department').agg(
    Student_Count=('Student_ID', 'count'),
    Average_Final_Score=('Final_Score', 'mean'),
    Max_Final_Score=('Final_Score', 'max'),
    Min_Final_Score=('Final_Score', 'min'))

dept_summary_sorted=dept_summary.sort_values(by='Average_Final_Score', ascending=False)
dept_summary_sorted
```

**Q15. NumPy and Pandas Integration**

```python
final_scores = df_sp['Final_Score'].to_numpy()

mean_score = np.mean(final_scores)
print("Mean (Average) Final Score:", mean_score)
print("Median Final Score:", np.median(final_scores))
print("Standard Deviation of Final Scores:", np.std(final_scores))
print("25th Percentile (Q1) Final Score:", np.percentile(final_scores, 25))
print("75th Percentile (Q3) Final Score:", np.percentile(final_scores, 75))
above_mean_students = df_sp[df_sp['Final_Score'] > mean_score]
print(f"\nNumber of students with a Final Score above the Mean: {len(above_mean_students)}")
print("Details of students above the mean score:")
print(above_mean_students[['Name', 'Final_Score']])
```

**SECTION C**

**Q16. Student Performance Analysis**

```python
#A Data Cleaning
df_sp = pd.read_csv('/content/drive/MyDrive/DSA/cases study/numoy panda/student_performance.csv')
print("Missing values in the df")
print(df_sp.isna().sum())
df_sp['Age'] = df_sp['Age'].fillna(df_sp['Age'].mean())
df_sp.head()
```

```python
df_sp['Average_Score'] = df_sp[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1)
df_sp['Pass_Status'] = np.where(df_sp['Average_Score'] >= 50, 'Pass', 'Fail')
df_sp.head()
```

```python
# 1. Top 5 Students by Average Score
top5 = df_sp.nlargest(5, 'Average_Score')

# 2. Department with Highest Average Score
highest_dept = df_sp.groupby('Department')['Average_Score'].mean().idxmax()

# 3. Average Attendance by Department
dept_att = df_sp.groupby('Department')['Attendance'].mean()

# 4. Pass/Fail Counts
pass_fail_counts = df_sp['Pass_Status'].value_counts()

print("\n--- Top 5 Students by Average Score ---")
print(top5[['Name', 'Average_Score']])

print("\n--- Department with Highest Average Score ---")
print(highest_dept)

print("\n--- Average Attendance by Department ---")
print(dept_att)

print("\n--- Pass/Fail Counts ---")
print(pass_fail_counts)
```

**Q18. Advanced NumPy and Pandas Analysis**

```python
df_sales = pd.read_csv('/content/drive/MyDrive/DSA/cases study/numoy panda/sales.csv')
df_sales.head()
```

```python
df_sales['Sales'] = pd.to_numeric(df_sales['Sales'], errors='coerce')
df_sales['Profit'] = pd.to_numeric(df_sales['Profit'], errors='coerce')
df_sales['Discount'] = pd.to_numeric(df_sales['Discount'], errors='coerce')

df_sales['Sales'] = df_sales['Sales'].fillna(0)
df_sales['Profit'] = df_sales['Profit'].fillna(0)
df_sales['Discount'] = df_sales['Discount'].fillna(0)

total_sales = df_sales['Sales'].sum()
total_profit = df_sales['Profit'].sum()
avg_sales = df_sales['Sales'].mean()

top_products = df_sales.groupby('Product')['Sales'].sum().nlargest(5)
top_region = df_sales.groupby('Region')['Sales'].sum().idxmax()

print(f"Total Sales: {total_sales:.2f}")
print(f"Total Profit: {total_profit:.2f}")
print(f"Average Sales: {avg_sales:.2f}")
print(f"Top Region by Sales: {top_region}")
```
